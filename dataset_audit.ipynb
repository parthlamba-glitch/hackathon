```python
import pandas as pd
import numpy as np


# ============================================================
#                    LOAD DATASET
# ============================================================

df = pd.read_csv(
    "dataset.csv"
)


# ============================================================
#                    BASIC INFO
# ============================================================

print()
print("=" * 60)
print("                PULSEGUARD DATASET AUDIT")
print("=" * 60)

print()

print(
    "Dataset shape:",
    df.shape
)


# ============================================================
#                    CLASS BALANCE
# ============================================================

print()
print("=" * 60)
print("                  CLASS BALANCE")
print("=" * 60)

print()

print(
    df["label"].value_counts()
)

print()

print(
    "REAL:",
    int(
        (df["label"] == 0).sum()
    )
)

print(
    "FAKE:",
    int(
        (df["label"] == 1).sum()
    )
)


# ============================================================
#                    MISSING VALUES
# ============================================================

print()
print("=" * 60)
print("                  MISSING VALUES")
print("=" * 60)

print()

missing = (
    df.isna().sum()
)

print(
    missing[
        missing > 0
    ]
)


# ============================================================
#                    INFINITE VALUES
# ============================================================

print()
print("=" * 60)
print("                 INFINITE VALUES")
print("=" * 60)

print()

numeric_df = df.select_dtypes(
    include=np.number
)

infinite = (
    np.isinf(
        numeric_df
    ).sum()
)

print(
    infinite[
        infinite > 0
    ]
)


# ============================================================
#                    DATA TYPES
# ============================================================

print()
print("=" * 60)
print("                    DATA TYPES")
print("=" * 60)

print()

print(
    df.dtypes
)


# ============================================================
#                    STATISTICS
# ============================================================

print()
print("=" * 60)
print("                NUMERIC SUMMARY")
print("=" * 60)

print()

print(
    numeric_df.describe().T
)


# ============================================================
#                    REAL vs FAKE
# ============================================================

print()
print("=" * 60)
print("              REAL vs FAKE SUMMARY")
print("=" * 60)

print()

features_to_check = [

    "face_detection_rate",

    "median_bpm",

    "mean_bpm",

    "mean_peak_strength",

    "temporal_variability",

    "forehead_left_corr",

    "forehead_right_corr",

    "left_right_corr",

    "regional_coherence"

]


available_features = [

    feature

    for feature
    in features_to_check

    if feature in df.columns
]


print(
    df.groupby(
        "label"
    )[available_features].mean()
)
```
